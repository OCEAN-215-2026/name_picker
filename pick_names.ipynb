```python
import math, random, html, uuid
from IPython.display import HTML, display
```

```python
with open("student_names.txt", "r") as infile:
    students = [line.strip() for line in infile]
```

```python
def pick_name(names, transition_time=0.1, spin_range=(3.0, 6.0), seed=None):

    "Animate a random name picker in a Jupyter Notebook output cell."

    if seed is not None:
        random.seed(seed)
    
    min_frames = max(1, int(spin_range[0] // transition_time))
    max_frames = int(spin_range[1] // transition_time)
    if min_frames > max_frames:
        min_frames, max_frames = max_frames, min_frames
    steps = random.randint(min_frames, max_frames)

    names = list(set([str(name) for name in names]))
    random.shuffle(names)
    
    if len(names) == 1:
        steps = 1
    k = (steps - 1) // len(names)
    
    frames = []
    for _ in range(k):
        batch = random.sample(names, len(names))
        if frames and batch[0] == frames[-1]:
            j = random.randrange(1, len(batch))
            batch[0], batch[j] = batch[j], batch[0]
        frames.extend(batch)

    remaining = steps - k * len(names)
    if frames and remaining == 1:
        batch = random.sample([name for name in names if name != frames[-1]], 1)
    else:
        batch = random.sample(names, remaining)
        if frames and batch[0] == frames[-1]:
            j = random.randrange(1, len(batch))
            batch[0], batch[j] = batch[j], batch[0]

    frames.extend(batch)
    winner = frames[-1]

    uid = "picker_" + uuid.uuid4().hex
    line_height = 4.0
    keyframes = "\n".join(
        f"{100 * i / steps:.6f}% {{ transform: translateY(-{i * line_height:.2f}rem); }}"
        for i in range(steps)
    )
    keyframes += f"\n100% {{ transform: translateY(-{(steps - 1) * line_height:.2f}rem); }}"
    items = "".join(f"<div class='item'>{html.escape(name)}</div>" for name in frames)
    duration_ms = int(steps * transition_time * 1000)

    display(HTML(f"""
    <style>
      #{uid} {{ font-family: system-ui, sans-serif; max-width: 28rem; text-align: center; }}
      #{uid} .window {{ height: {line_height}rem; overflow: hidden; border: 2px solid #805ad5;
                          border-radius: 12px; background: #32006e; color: #cec5af; }}
      #{uid} .item {{ height: {line_height}rem; line-height: {line_height}rem;
                      font-size: 1.5rem; font-weight: 700; white-space: nowrap;
                      overflow: hidden; text-overflow: ellipsis; }}
      #{uid} .reel {{ animation: {uid}_spin {duration_ms}ms steps(1, end) forwards; }}
      #{uid} .result {{ visibility: hidden; margin-top: .7rem; font-size: 1.1rem;
                        animation: {uid}_show 0s linear {duration_ms}ms forwards; }}
      @keyframes {uid}_spin {{ {keyframes} }}
      @keyframes {uid}_show {{ to {{ visibility: visible; }} }}
    </style>
    <div id="{uid}">
      <div class="window"><div class="reel">{items}</div></div>
      <div class="result">🎉 Winner: <strong>{html.escape(winner)}</strong></div>
    </div>
    """))
```

```python
pick_names(students)
```
